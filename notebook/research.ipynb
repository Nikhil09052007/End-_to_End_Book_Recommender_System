```python
# Now join ratings with books

ratings_with_books = ratings.merge(books, on='ISBN')
```

```python
ratings_with_books.head()
```

```python

number_rating = ratings_with_books.groupby('title')['rating'].count().reset_index()
```

```python

number_rating.head()
```

```python

number_rating.rename(columns={'rating':'num_of_rating'},inplace=True)
```

```python
number_rating.head()
```

```python

final_rating = ratings_with_books.merge(number_rating, on='title')
```

```python
final_rating.head()
```

```python
# Lets take those books which got at least 50 rating of user

final_rating = final_rating[final_rating['num_of_rating'] >= 50]
```

```python
final_rating.head()
```

```python
# Since I have duplicate values in the final_rating dataset, I will drop them and keep the first one. This is because I want to keep only one record of each book for my analysis.
final_rating.drop_duplicates(['user_id','title'],inplace=True)
```

```python
# checking for missing values in the final_rating dataset. I will drop those records which have missing values as they are not useful for my analysis.  
final_rating.isnull().sum()
```

```python
# For clustering and to find the similar books, I will use the pivot table to create a matrix of users and books. The rows will be the book titles and the columns will be the user ids. The values will be the ratings given by the users to the books. If a user has not rated a book, the value will be 0. This matrix will be used to find the similar books using cosine similarity.
book_pivot = final_rating.pivot(index='title', columns='user_id', values='rating').fillna(0)
```

```python
book_pivot.head()
```

```python
book_pivot.shape
```

```python
# filling nan with zeroes as we are not interested in the books which are not rated by the users. This will help us to find the similar books using cosine similarity. 
book_pivot.fillna(0, inplace=True)
```

```python
from scipy.sparse import csr_matrix
```

```python
book_sparse = csr_matrix(book_pivot)
```

```python
book_sparse
```

```python
type(book_sparse)
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(algorithm='brute', metric='cosine')
```

```python
model.fit(book_sparse)
```
